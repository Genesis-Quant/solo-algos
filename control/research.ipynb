# Control 研究

通过插件将上游项目安装为 dev 依赖。只在此 Notebook 组合上游参数，算法包不依赖上游项目。

```python
from importlib import import_module
from control import ControlAlgo, ControlReportForm
from scheme import StockPool, algo_options
from scheme.base import ControlAnalysisParams

model_options = algo_options("model")
optimize_options = algo_options("optimize")

model_options, optimize_options
```

```python
model_entry = next(iter(model_options))
ModelParams = import_module(model_entry.split(":")[0]).ModelParams
model_params = ModelParams()  # 在此设置该上游算法参数

optimize_entry = next(iter(optimize_options))
OptimizeParams = import_module(optimize_entry.split(":")[0]).OptimizeParams
optimize_params = OptimizeParams()  # 在此设置该上游算法参数

form = ControlReportForm(
    start="2026-06-01",
    end="2026-06-06",
    pool=StockPool.SSE50,
    model=model_entry,  # 查看上一格选项，可改为指定的入口字符串
    optimize=optimize_entry,  # 查看上一格选项，可改为指定的入口字符串
)
params = ControlAnalysisParams(
    **form.build().model_dump(),
    **model_params.model_dump(),
    **optimize_params.model_dump(),
)
params
```

```python
report = params.run(ControlAlgo)
report.show()
```
